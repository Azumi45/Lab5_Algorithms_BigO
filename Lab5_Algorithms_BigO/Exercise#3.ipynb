```python
LAST_NAME = "Celo"
STUDENT_ID = "TUPM-26-1941"
FAVORITE_ARTIST = "Felip"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(char.upper()) - 64 for char in FAVORITE_ARTIST if char.isalpha())
```

```python
# Generated Order Dataset
order = [
    (ID_SUM * SEED_DIGIT) % 100, (SEED_DIGIT * 3 * 10) % 100, (ID_SUM * NAME_LENGTH) % 100, (ID_SUM - SEED_DIGIT) % 100,
        ((5 - SEED_DIGIT) * 10) % 100, 
        (NAME_LENGTH - (- ID_SUM)) % 100,
        (ID_SUM - NAME_LENGTH) % 100, 
        ((SEED_DIGIT + NAME_LENGTH) * 10) % 100, (NAME_LENGTH * 11 * SEED_DIGIT) % 100
]
```

```python
import time

def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials
```

```python
# BUBBLE SORT

print("Random:", order)
def bubble_sort(order):
    a, comparisons, swaps = order.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

order_random, comparisons, swaps = bubble_sort(order)

print("Comparisons:", comparisons)
print("Swaps:", swaps)

bubble_result, bubble_time = measure(bubble_sort, order)
print(f"Bubble Sort: {bubble_time:.8f} s")
```

```python
# INSERTION SORT
def insertion_sort(order):
    a, comparisons, shifts = order.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

order_sorted, comparisons, shifts = insertion_sort(order)

print("Sorted:", order_sorted)
print("Comparisons:", comparisons)
print("Shifts:", shifts)

insertion_result, insertion_time = measure(insertion_sort, order)
print(f"Insertion Sort: {insertion_time:.8f} s")
```

```python
# SELECTION SORT
def selection_sort(order):
    a, comparisons, swaps = order.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] > a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

order_reversed, comparisons, swaps = selection_sort(order)

print("Reversed Ordered:", order_reversed)
print("Comparisons:", comparisons)
print("Swaps:", swaps)

selection_result, selection_time = measure(selection_sort, order)
print(f"Selection Sort: {selection_time:.8f} s")
```
