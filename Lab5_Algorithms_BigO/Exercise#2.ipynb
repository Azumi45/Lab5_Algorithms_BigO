```python
LAST_NAME = "Celo"
STUDENT_ID = "TUPM-26-1941"
FAVORITE_ARTIST = "Felip"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(char.upper()) - 64 for char in FAVORITE_ARTIST if char.isalpha())
```

```python
data_a = [
    (ID_SUM * SEED_DIGIT) % 100, (SEED_DIGIT * 3) % 100, 
    (NAME_LENGTH + SEED_DIGIT) % 100, (ID_SUM * NAME_LENGTH) % 100, (SEED_DIGIT - 1) % 100 
]

data_b = [
    (ID_SUM - SEED_DIGIT) % 100,
    (5 - SEED_DIGIT) % 100, 
    (NAME_LENGTH - (- ID_SUM)) % 100,
    (ID_SUM - NAME_LENGTH) % 100, 
    (SEED_DIGIT + NAME_LENGTH) % 100, 
]

print(f"Department A Dataset: {data_a}")

# Number of Records B
def search_case(data, data_a):
    for i, value in enumerate(data, 1):
        if value == data_a:
            return i 
    return len(data)

for record, target in {"Number of Records": 0}.items():
    print(f"{record}: {search_case(data, data_a)}")

print(f"Department B Dataset: {data_b}")

# Number of Records B
def search_case(data, data_b):
    for i, value in enumerate(data, 1):
        if value == data_b:
            return i 
    return len(data)

for record, target in {"Number of Records": 5}.items():
    print(f"{record}: {search_case(data, data_b)}")
```

```python
# Department A Dataset
def insertion_sort(data_a):
    a, comparisons, shifts = data_a.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

sorted_data, comparisons, shifts = insertion_sort(data_a)

print("Department A Dataset:", sorted_data)
```
