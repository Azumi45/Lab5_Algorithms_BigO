```python
LAST_NAME = "Celo"
STUDENT_ID = "TUPM-26-1941"
FAVORITE_ARTIST = "Felip"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(char.upper()) - 64 for char in FAVORITE_ARTIST if char.isalpha())

# print(f"{}")
```

```python
# 1. Generate the original and rotated incident records

incident_records = [
    (ID_SUM * SEED_DIGIT) % 100,
    (SEED_DIGIT * 5) % 100, 
    (NAME_LENGTH + ID_SUM) % 100,
    (ID_SUM * NAME_LENGTH) % 100, 
    (SEED_DIGIT - NAME_LENGTH) % 100, 
]

target_incident =  ARTIST_SUM - ID_SUM - 2
rotation_point = SEED_DIGIT
rotation_records = incident_records[rotation_point:] + incident_records[:rotation_point]

print(f"Original Dataset: {incident_records}")
print(f"Rotated Dataset: {rotation_records}")
print(f"Target Incident: {target_incident}")

```

```python
# 2. Search for the target incident
# 3. Report its position or indicate that it is found or not
# 4. Record th number of comparisons

if target_incident in rotation_records:
    target_index = rotation_records.index(target_incident)
    comparisons = target_index + 1
    print(f"Target Index:({target_incident}) | Found at index {target_index}")
else:
    comparisons = len(rotation_records)
    print(f"Target Incident not Found")

print(f"Comparison Counts: {comparisons}")
```

```python
# 5. Compare the results with Linear Search
def linear_search(rotation_records, target_incident):
    for i, value in enumerate(incident_records):
        if value == target_incident:
            return i, i + 1
    return -1, len(rotation_records)

index, comparisons = linear_search(rotation_records, target_incident)

print("Target:", target_incident)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# Determine the Time and Space Complexity

incident_records = [
    (ID_SUM * SEED_DIGIT) % 100,
    (SEED_DIGIT * 5) % 100, 
    (NAME_LENGTH + ID_SUM) % 100,
    (ID_SUM * NAME_LENGTH) % 100, 
    (SEED_DIGIT - NAME_LENGTH) % 100, 
]

# Time Complexity
print(f"===== Time Complexity ======")

def linear_operations(n):
    count = 0
    for _ in range(n):
        count += 1
    return count

for n in incident_records:
    print(f"n = {n:5d} | operations = {linear_operations(n):5d}")

# Space Complexity
print(f"===== Space Complexity ======")
def constant_space(n):
    total = 0
    for value in range(n):
        total += value
    return total

def linear_space(n):
    return list(range(n))

for n in incident_records:
    data = linear_space(n)
    print(f"n={n:5d} | 0(1): constant | 0(n): {len(data)}")


```
